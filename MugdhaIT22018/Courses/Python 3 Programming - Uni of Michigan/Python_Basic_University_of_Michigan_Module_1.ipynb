## Module overview
This module introduces Python programming, expressions, data types, variables, functions, user input, turtle graphics, loops, modules, and debugging.

## 1. Introduction to Python
- Python is a beginner-friendly, general-purpose programming language.
- It is used in data science, AI, automation, web development, and digital art.
- A program is a set of instructions executed by the Python interpreter.
- Use `print()` to display information. An expression can run without displaying anything.

## 2. Expressions, values, and operators
- An **expression** is code that Python evaluates to produce a value.
- Common value types: `int` (whole numbers), `float` (decimal numbers), and `str` (text).
- Arithmetic operators: `+`, `-`, `*`, `/`, `//` (floor division), `%` (remainder), and `**` (power).
- Order of operations: parentheses, exponentiation, multiplication/division, then addition/subtraction.

## 3. Functions
- A function accepts arguments, performs an operation, and may return a value.
- Call a function using its name followed by parentheses: `len("Python")`.
- Nested function calls are evaluated from the inside out. Arguments are evaluated from left to right.

## 4. Data types and type conversion
- Use `type(value)` to identify a value's type.
- Strings can use single, double, or triple quotes.
- `int()`, `float()`, and `str()` convert values between common types.
- Converting a float to an integer truncates the decimal part; it does not round.

## 5. Variables and assignment
- A variable is a name that refers to a stored value.
- Assign values with `=` and update them with reassignment.
- Variable names may contain letters, digits, and underscores, but cannot start with a digit.
- Python is case-sensitive: `score` and `Score` are different names.

## 6. Statements and evaluation
- An expression produces a value; a statement is an instruction executed by Python.
- In an assignment such as `total = 1 + 2`, Python evaluates the right side first, then assigns the result.
- Complex expressions are evaluated by following operator precedence and function-call order.

## 7. Reassignment and shortcut operators
- `x = x + 1` updates the value of `x`.
- Shortcuts include `+=`, `-=`, `*=`, and `/=`.
- The right-hand side is evaluated before the variable is updated.

## 8. Avoid hard-coding
- Hard-coding means writing a fixed answer directly into a program.
- It makes programs inflexible when input values change.
- Calculate results from variables instead, so the program works with different values.

## 9. User input
- `input()` displays a prompt and always returns a string.
- Convert input with `int()` or `float()` before doing numeric calculations.
- Invalid input can cause a `ValueError`; validate input when needed.

## 10. Turtle graphics, objects, and loops
- The `turtle` module controls an on-screen turtle that can draw shapes.
- Create objects such as `turtle.Screen()` and `turtle.Turtle()`.
- Use dot notation to call methods, e.g. `alex.forward(100)`.
- A `for` loop repeats an indented block. `range(n)` produces `n` iterations.
- Turtle methods include `forward()`, `left()`, `right()`, `penup()`, `pendown()`, and `stamp()`.

## 11. Modules and imports
- A module is a Python file or library containing reusable code.
- `import random` lets you call `random.randrange(1, 10)`.
- `from math import sqrt` imports a specific function for direct use.

## 12. Debugging
Three common error types:
- **Syntax error:** code breaks Python's grammar, such as a missing parenthesis.
- **Runtime error:** occurs while the program runs, such as division by zero.
- **Semantic (logic) error:** the program runs but gives an incorrect result.

Debugging tips:
- Read the error message and check the reported line and nearby lines.
- Check spelling, capitalization, variable names, and function arguments.
- Build programs incrementally: write a small part, run it, test it, then continue.
- Aim to “get something working and keep it working.”

---
**Key takeaway:** Understand the values your expressions produce, store useful results in variables, use functions and loops to avoid repetition, accept and convert user input carefully, and debug programs step by step.


## Practical code examples
Run the following cells to practise the main ideas from Module 1.

```python
# 1. Printing, expressions, and arithmetic operators
print("Hello, Python!")

print(100 + 200)   # Addition
print(10 - 3)      # Subtraction
print(6 * 7)       # Multiplication
print(7 / 2)       # Division
print(7 // 2)      # Floor division
print(7 % 2)       # Remainder
print(4 ** 2)      # Exponentiation
```

```python
# 2. Checking data types
print(type(100))          # int
print(type(3.14))         # float
print(type("hello"))      # str
print(type("100"))        # str, because it is inside quotes
```

```python
# 3. Function calls and nested expressions
def square(number):
    return number ** 2

def add(first, second):
    return first + second

print(square(4))                  # 16
print(add(5, 3))                   # 8
print(square(3) + 2)               # 11
print(add(square(2), square(3)))   # 13
```

```python
# 4. Type conversion
decimal_value = 3.999
print(int(decimal_value))       # 3 (truncates; does not round)
print(float("12.5"))            # 12.5
print(str(25) + " students")    # 25 students
```

```python
# 5. Variables, reassignment, and shortcut operators
score = 10
print("Starting score:", score)

score = score + 5
print("After reassignment:", score)

score += 3
score -= 2
print("Final score:", score)
```

```python
# 6. Avoiding hard-coded results
# Calculate the average from variables instead of writing a fixed answer.
mark1 = 80
mark2 = 90
average = (mark1 + mark2) / 2

print("Average mark:", average)
```

```python
# 7. User input and numeric conversion
# Run this cell and enter a name when prompted.
name = input("Enter your name: ")
print("Hello,", name)

# input() returns a string. Convert it before numerical calculations.
age_text = input("Enter your age: ")
age = int(age_text)
print("Next year you will be", age + 1)
```

```python
# 8. Convert seconds into hours, minutes, and seconds
total_seconds = 3672

hours = total_seconds // 3600
remaining_seconds = total_seconds % 3600
minutes = remaining_seconds // 60
seconds = remaining_seconds % 60

print(hours, "hours,", minutes, "minutes,", seconds, "seconds")
```

```python
# 9. for loop and range()
for number in range(1, 6):
    print("Count:", number)

# A loop can also calculate a total.
total = 0
for number in range(1, 6):
    total += number

print("Sum:", total)
```

```python
# 10. Importing modules
import random
import math

print("Random number:", random.randrange(1, 11))
print("Square root of 81:", math.sqrt(81))
```

```python
# 11. Debugging practice: identify and fix common mistakes

# NameError example (fixed):
student_name = "Mugdha"
print(student_name)  # Use the variable name that was actually defined.

# ZeroDivisionError prevention:
numerator = 10
denominator = 0

if denominator != 0:
    print(numerator / denominator)
else:
    print("Cannot divide by zero.")

# Semantic error example (fixed):
length = 5
width = 4
area = length * width  # Correct formula for rectangle area
print("Rectangle area:", area)
```

## Turtle graphics examples
**Note:** Turtle opens a separate graphics window and may not work in every hosted notebook environment. Run these examples in a local Python installation that supports the `turtle` module.

```python
# 12. Draw a square with Turtle
import turtle

screen = turtle.Screen()
alex = turtle.Turtle()

for _ in range(4):
    alex.forward(100)
    alex.left(90)

screen.exitonclick()
```

```python
# 13. Draw a simple spiral with Turtle
import turtle

screen = turtle.Screen()
elan = turtle.Turtle()
elan.color("blue")
elan.shape("turtle")

for distance in range(10, 160, 10):
    elan.forward(distance)
    elan.right(90)

screen.exitonclick()
```
