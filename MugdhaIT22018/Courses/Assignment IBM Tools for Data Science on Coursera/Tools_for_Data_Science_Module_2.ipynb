**IBM / Coursera | Quick Revision Notes**

**Topic:** Languages of Data Science  
**Goal:** Choose languages by task and remember their main features, libraries, and examples.


# 1. Choosing a Data Science Language

- Choose based on **problem, job role, and organization/project needs**.
- **Start with Python, R, and SQL**.
- Other relevant languages: **Scala, Java, C++, JavaScript, Julia** (also PHP, Go, Ruby, Visual Basic for specialized uses).
- Roles include **Data Scientist, Data Engineer, Statistician, Business Analyst, Product Manager**.


# 2. Python — General-Purpose Data Science

- **Open-source**, beginner-friendly, readable, versatile, and widely used.
- Uses: **data analysis, AI/ML, web development, IoT, NLP**.

| Purpose | Key Python libraries |
|---|---|
| Data manipulation | **Pandas** |
| Numerical/scientific computing | **NumPy, SciPy** |
| Visualization | **Matplotlib** |
| Machine learning | **Scikit-learn** |
| Deep learning | **TensorFlow, PyTorch, Keras** |
| Natural language processing | **NLTK** |

**Community:** Python Software Foundation; PyLadies supports inclusion.


# 3. R — Statistics and Data Analysis

- **Free software** popular in statistics, research, and data mining.
- **Array-oriented syntax** makes mathematical operations convenient.
- Large package ecosystem; interoperates with **Python, C++, Java**.
- Communities: **useR, R-Ladies**.
- **Free software vs. open source:** Both may share licenses such as GPL; FSF emphasizes user freedoms, while OSI emphasizes open-source principles and practical development.


# 4. SQL — Relational Databases

- **SQL = Structured Query Language**.
- A **non-procedural/declarative** language: specify *what data* you want, rather than the step-by-step procedure.
- Designed to query and manage **relational tables (rows and columns)**.
- **ANSI-standardized**; skills transfer across **MySQL, PostgreSQL, Oracle**, etc.
- SQL-style interfaces are also found in some **NoSQL and big-data** systems.


# 5. Other Important Languages

| Language | Main strength / use | Tools or examples from the course |
|---|---|---|
| **Java** | Object-oriented, scalable applications | Weka, Java-ML, Deeplearning4j, Hadoop |
| **Scala** | JVM + functional programming; big-data processing | Apache Spark, MLlib, GraphX, Spark Streaming |
| **C++** | High performance and low-level control | Underlying implementations in TensorFlow, MongoDB, Caffe |
| **JavaScript** | Browser/server applications; ML on the web | TensorFlow.js, Node.js, R-js |
| **Julia** | Fast numerical/scientific computing; parallelism | JuliaDB |

**Remember:** JavaScript is a different language from Java.


# 6. Quick Comparison

| Language | Best remembered for |
|---|---|
| **Python** | General-purpose data science, ML, rich libraries |
| **R** | Statistics, data analysis, research |
| **SQL** | Querying and managing relational data |
| **Scala** | Apache Spark and big data |
| **Java** | Scalable, object-oriented applications |
| **C++** | Performance-critical systems |
| **JavaScript** | Web-based machine learning |
| **Julia** | High-performance numerical computing |


# 7. Quiz / Exam Flash Review

1. **Top beginner languages?** Python, R, SQL.
2. **SQL full form?** Structured Query Language.
3. **What kind of language is SQL?** Non-procedural / declarative.
4. **Python library for NLP?** NLTK.
5. **Python libraries for scientific computing?** NumPy, SciPy, Pandas.
6. **Popular R feature?** Array-oriented syntax and statistical packages.
7. **Scala-based big-data engine?** Apache Spark.
8. **Java-based ML examples?** Weka, Java-ML, Deeplearning4j.
9. **JavaScript ML library?** TensorFlow.js.
10. **Julia data tool mentioned in course?** JuliaDB.

**One-line takeaway:** Pick the language for the task: **Python for broad data science, R for statistics, SQL for databases**, and specialized languages for scale, speed, or web use.
