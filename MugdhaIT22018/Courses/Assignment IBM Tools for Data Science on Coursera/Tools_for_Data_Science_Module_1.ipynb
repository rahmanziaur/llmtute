# Tools for Data Science — Module 1

**IBM / Coursera | Concise Study Notes**

**Focus:** Categories of data science tools, open-source tools, commercial tools, cloud tools, and development environments.


# 1. Data Science Workflow

**Typical workflow:** Collect/store data → Integrate & transform (ETL) → Visualize/explore → Build model → Deploy → Monitor & assess.

| Task | Purpose | Examples |
|---|---|---|
| Data management | Store, organize and retrieve data | MySQL, PostgreSQL, MongoDB |
| Data integration & transformation | Combine, clean and move data (ETL/ELT) | Airflow, Kafka, NiFi |
| Visualization | Communicate patterns using charts/dashboards | Kibana, Superset, Tableau |
| Model building | Train predictive models | Watson Studio, SPSS Modeler |
| Deployment | Make models available for applications | Seldon, TensorFlow Serving |
| Monitoring & assessment | Track performance, bias and explainability | Prometheus, Watson OpenScale |
| Code asset management | Version control and collaboration | Git, GitHub, GitLab |
| Data asset management | Metadata, governance and access control | Apache Atlas, Egeria |
| Development/execution | Write, test and run code | Jupyter, RStudio, Spark |


# 2. Open-Source Tools — Key Examples

| Category | Tools and key uses |
|---|---|
| Relational databases | **MySQL, PostgreSQL:** SQL-based structured data |
| NoSQL databases | **MongoDB, CouchDB:** document data; **Cassandra:** distributed NoSQL |
| Storage/search | **HDFS:** distributed file storage; **Ceph:** unified storage; **Elasticsearch:** indexed search |
| Data pipelines | **Airflow:** workflow scheduling; **Kafka:** event streaming; **NiFi:** visual data flows; **Kubeflow:** ML pipelines |
| Processing/integration | **Spark SQL:** structured large-scale data; **Node-RED:** visual event flows |
| Visualization | **Kibana:** Elasticsearch dashboards; **Superset:** BI dashboards; **Hue:** Hadoop interface; **PixieDust:** notebook visualization |
| Model deployment | **Seldon:** Kubernetes model serving; **MLeap:** portable models; **TensorFlow Serving:** model APIs; **TensorFlow Lite:** mobile/embedded inference |
| Monitoring | **ModelDB:** experiment tracking; **Prometheus:** metrics/alerts; **AI Fairness 360:** bias; **AI Explainability 360:** explanations; **Adversarial Robustness 360:** attack robustness |
| Code collaboration | **Git:** version control; **GitHub, GitLab, Bitbucket:** hosting and collaboration |
| Data governance | **Apache Atlas, ODPi Egeria, Kylo:** metadata and lineage |


# 3. Development & Execution Environments

- **Jupyter Notebook:** Interactive notebook combining code, text, equations and plots; supports multiple languages via kernels.
- **JupyterLab:** More flexible workspace with notebooks, editors and terminals.
- **RStudio:** IDE for R programming and statistical analysis.
- **Spyder:** Scientific Python IDE; **PyCharm:** Python IDE; **Visual Studio:** multi-language IDE.
- **Apache Zeppelin:** Notebook with built-in plotting.
- **Apache Spark:** Scalable batch and stream data processing; **Apache Flink:** stream-focused processing; **Ray:** distributed computing and ML workloads.
- **KNIME / Orange:** Visual, drag-and-drop data science tools; Orange is simpler.
- **Anaconda Navigator:** GUI for managing Python tools and environments.

# 5. Cloud-Based Tools

- **Cloud databases:** Amazon DynamoDB, IBM Db2 as a Service.
- **Cloud integration:** Informatica Cloud Data Integration, IBM Data Refinery.
- **Cloud model training/deployment:** IBM Watson Machine Learning, Google AI Platform Training (name used in the course notes).
- **Cloud monitoring:** IBM Watson OpenScale, Amazon SageMaker Model Monitor.
- **SaaS:** Provider manages much of the service infrastructure, such as maintenance and updates.
- **REST API:** Common way for applications to access deployed models and services.


# 6. IBM Model Development Tools

- **IBM Watson Studio:** Collaborative environment for data preparation and model development.
- **IBM AutoAI:** Automates algorithm and hyperparameter exploration.
- **IBM Watson Machine Learning:** Cloud training and deployment of ML models.
- **IBM Watson OpenScale:** Monitors model performance, fairness, explainability and drift.


# 7. Quick Comparisons

| Concept | Remember |
|---|---|
| SQL vs NoSQL | SQL relational tables vs flexible/non-relational models |
| ETL vs ELT | Transform before loading vs load before transforming |
| Jupyter Notebook vs JupyterLab | Notebook document vs integrated multi-panel workspace |
| Git vs GitHub | Version control system vs online Git hosting/collaboration |
| Model building vs deployment | Train model vs make it available for use |
| Monitoring vs explainability | Track behavior/metrics vs explain predictions |
| Open-source vs commercial vs cloud | Licensing model vs vendor offering vs delivery/hosting model; categories can overlap |


# 8. Quiz Revision — Remember These

1. **ETL = Extract, Transform, Load.**
2. **MySQL / PostgreSQL** = relational databases; **MongoDB / Cassandra** = NoSQL.
3. **Airflow** = workflow orchestration; **Kafka** = streaming; **NiFi** = data flows.
4. **Kibana** works closely with **Elasticsearch**; **Superset** builds dashboards.
5. **TensorFlow Lite** targets mobile/embedded devices; **TensorFlow Serving** serves trained models.
6. **Prometheus** monitors metrics; **ModelDB** tracks experiments.
7. **Git** tracks revisions; **GitHub** supports repositories and collaboration.
8. **Jupyter** combines executable code, narrative text and visualizations.
9. **Watson AutoAI** automates model development; **OpenScale** monitors deployed models.
10. **Cloud tools** can provide managed storage, development, deployment and monitoring.
