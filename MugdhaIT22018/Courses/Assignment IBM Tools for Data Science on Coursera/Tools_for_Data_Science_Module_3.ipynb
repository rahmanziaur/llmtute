## 1. Libraries for Data Science
| Category | Tools | Main purpose |
|---|---|---|

**Other ecosystems:** Apache Spark processes data across clusters (Python/R/Scala/SQL); Scala has Vegas and BigDL; R has ggplot2 and deep-learning interfaces.

## 2. APIs and REST APIs
- **API (Application Programming Interface):** An interface that lets software components communicate without exposing internal implementation.
- **REST (Representational State Transfer):** An approach to accessing web resources using HTTP requests.
- **Client:** Sends requests; **resource:** Data/service; **endpoint:** URL for a resource.
- REST APIs commonly exchange **JSON** and use HTTP methods such as **GET** and **POST**.
- **Examples:** IBM Watson Speech-to-Text (audio → text) and Language Translator (text translation).
- TensorFlow provides APIs for several languages even though its core uses C++.

## 3. Datasets and Data Sources
- **Dataset:** Collection of data (tables/CSV, trees, graphs; text, images, audio, video).
- **Open dataset:** Shared for reuse under specified terms; useful for learning and research.
- **Proprietary dataset:** Controlled by an owner; access or redistribution may be restricted.

| Source | Examples |
|---|---|
| Government | data.gov, data.un.org |
| Finance | World Bank Data, FRED |
| General | Kaggle Datasets |
| Academic/business | NCES, Yelp Dataset |
| Earth/space | NASA Earthdata |

**Always check the dataset license before using or redistributing data.**

## 4. Dataset Licenses — Quick Reference
| License / term | Key condition |
|---|---|
| Public domain / PDDL | Broad reuse permitted |
| CC BY / ODC-BY | Give attribution |
| CC BY-SA / ODbL | Attribution and share-alike obligations |
| CC BY-NC | Noncommercial use only |
| CC BY-ND | No derivatives shared |
| CDLA-Permissive | Permissive data sharing with required notices |
| CDLA-Sharing | Sharing obligations for redistributed data |

**CDLA = Community Data License Agreement.** Exact obligations vary by license/version.

## 5. Machine Learning Models
**Machine learning:** Train models to learn patterns from data and make predictions.

| Type | Training data / method | Typical task |
|---|---|---|
| **Supervised** | Labeled examples | Regression (numbers), classification (classes) |
| **Unsupervised** | Unlabeled data | Clustering, anomaly detection |
| **Reinforcement** | Trial, error and rewards | Learning actions/policies |

**Deep learning:** Subfield of ML using neural networks for tasks involving images, language and audio. Frameworks include TensorFlow, PyTorch and Keras.

**Workflow:** Collect & prepare data → select model → train/evaluate → deploy.

## 6. IBM Data and Model Asset Exchanges
- **DAX (Data Asset eXchange):** IBM collection of open datasets.
- **MAX (Model Asset eXchange):** IBM repository of pre-trained deep-learning models and model-serving microservices.
- MAX use cases: image/object recognition, text classification, named entity recognition, audio/video analysis, human pose detection.
- A MAX microservice may include a trained model, input preprocessing, output postprocessing and an API.
- **Docker** packages services; **Kubernetes/OpenShift** helps deploy, scale and manage them.
- Repository mentioned in course: https://github.com/CODAIT/max-central-repo
