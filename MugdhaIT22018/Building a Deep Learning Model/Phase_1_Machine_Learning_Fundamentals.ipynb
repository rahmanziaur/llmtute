## 1. Artificial Intelligence (AI)

**Artificial Intelligence (AI)** is a field of computer science focused on creating systems that can perform tasks that normally require human intelligence.

Examples include:
- Face recognition
- Voice assistants
- Self-driving cars
- Chatbots
- Medical image analysis
- Fraud detection

AI is the broadest concept among AI, Machine Learning, and Deep Learning.


## 2. Machine Learning (ML)

**Machine Learning (ML)** is a subset of Artificial Intelligence. It enables computers to learn patterns and relationships from data and use those patterns to make predictions or decisions.

For example, a model can be trained using emails labeled as `Spam` or `Not Spam`. After learning from these examples, it can predict the category of a new email.

### Traditional Programming vs. Machine Learning

- **Traditional programming:** A programmer writes explicit rules, and the computer follows them.
- **Machine Learning:** The computer is given data and an algorithm learns patterns from that data.

Machine Learning performance depends on factors such as data quality, feature selection, model choice, and evaluation.


## 3. Deep Learning (DL)

**Deep Learning** is a subset of Machine Learning that uses multi-layered Artificial Neural Networks to learn complex patterns.

For example, in image recognition, different neural network layers may learn:
1. Basic edges and visual patterns.
2. Textures and shapes.
3. Object parts.
4. High-level representations used to identify the object.

### Relationship

`Artificial Intelligence ⊃ Machine Learning ⊃ Deep Learning`

- **AI:** The broad field of intelligent systems.
- **ML:** Learning patterns from data.
- **DL:** Learning complex representations using deep neural networks.

Deep Learning is widely used in image, audio, text, and other complex data tasks. It may require substantial data and computational resources.


## 4. Types of Machine Learning

### 4.1 Supervised Learning

In **Supervised Learning**, a model is trained using labeled data. Each input example is paired with a correct output or answer.

Example: Predicting whether a student will `Pass` or `Fail` based on study hours and attendance.

The two major supervised learning tasks are:
1. Regression
2. Classification

### 4.2 Unsupervised Learning

In **Unsupervised Learning**, the dataset does not contain target labels. The model attempts to discover patterns, similarities, or groups in the data.

Example: Grouping customers according to their shopping behavior. This is commonly called **Clustering**.

### 4.3 Semi-Supervised Learning

**Semi-Supervised Learning** uses a small amount of labeled data together with a larger amount of unlabeled data.

Example: Training with 100 labeled medical images and 10,000 unlabeled medical images.

### 4.4 Reinforcement Learning

In **Reinforcement Learning**, an agent interacts with an environment, takes actions, and learns from rewards or penalties.

Example: A game-playing agent learns which actions lead to better outcomes.

Important terms:
- **Agent:** The learner or decision-maker.
- **Environment:** The world in which the agent operates.
- **Action:** A decision made by the agent.
- **Reward:** Feedback received after an action.
- **Policy:** A strategy for choosing actions.


## 5. Regression vs. Classification

### 5.1 Regression

Regression is used when the target is a **continuous numerical value**.

Examples:
- House price prediction
- Salary prediction
- Temperature prediction
- Sales forecasting

A simple Linear Regression equation is:

\[
\hat{y} = mx + c
\]

Where:
- `x` = Input feature
- `ŷ` = Predicted output
- `m` = Slope or coefficient
- `c` = Intercept

For example, if the illustrative equation is `ŷ = 10x + 25`, then for `x = 5`:

\[
\hat{y} = 10(5) + 25 = 75
\]

In practice, the model learns the coefficient and intercept from training data.

### 5.2 Classification

Classification is used when the target is a **category or class**.

Examples:
- Spam / Not Spam
- Pass / Fail
- Cat / Dog
- Fraud / Legitimate

Two common forms are:
- **Binary Classification:** Two possible classes.
- **Multiclass Classification:** More than two possible classes.

### Comparison

| Property | Regression | Classification |
|---|---|---|
| Target | Continuous numerical value | Category or class |
| Example | House price | Spam or Not Spam |
| Common metrics | MAE, MSE, RMSE, R² | Accuracy, Precision, Recall, F1 |
| Example algorithm | Linear Regression | Logistic Regression |

**Important:** Despite its name, Logistic Regression is primarily used for classification.

## 7. Training, Validation, and Testing

A dataset is often divided into separate parts to evaluate whether a model can generalize to data it has not seen before.

### Training Set

The model learns patterns and estimates its parameters using the training data.

### Validation Set

The validation set is used to compare model configurations and tune hyperparameters. A separate validation set is not required in every workflow; cross-validation is another common approach.

### Test Set

The test set is reserved for final evaluation on unseen data. It should not be repeatedly used for model tuning or selection.

### Example: 1,000 Samples

A possible split is:
- 70% training = 700 samples
- 15% validation = 150 samples
- 15% testing = 150 samples

Another common approach is 80% training and 20% testing. The appropriate split depends on the dataset and task.


## 8. Overfitting and Underfitting

### 8.1 Overfitting

**Overfitting** occurs when a model performs very well on training data but performs poorly on unseen data. The model may learn noise or patterns that are specific to the training dataset.

Possible solutions:
- Use more representative training data.
- Apply regularization.
- Reduce model complexity.
- Use early stopping.
- Use dropout in neural networks.
- Apply data augmentation when appropriate.

### 8.2 Underfitting

**Underfitting** occurs when a model is too simple or insufficiently trained to capture important patterns in the training data. Its performance may be poor on both training and test data.

Possible solutions:
- Use a more suitable or expressive model.
- Improve the input features.
- Reduce excessive regularization.
- Train appropriately.

### Comparison

| Condition | Training Error | Test Error |
|---|---|---|
| Underfitting | High | High |
| Good Fit | Low | Low |
| Overfitting | Very low in many cases | Relatively high |

The goal is to build a model that learns useful patterns from training data and generalizes well to unseen data.


## 9. Model Evaluation Metrics

Evaluation metrics help measure how well a trained model performs. Regression and Classification use different metrics.

### 9.1 Regression Metrics

Let:
- `y` = Actual value
- `ŷ` = Predicted value
- `n` = Number of samples

**MAE — Mean Absolute Error**

\[
MAE = \frac{1}{n}\sum_{i=1}^{n}|y_i-\hat{y}_i|
\]

MAE is the average absolute difference between actual and predicted values.

**MSE — Mean Squared Error**

\[
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
\]

MSE squares the errors, giving larger errors a greater penalty.

**RMSE — Root Mean Squared Error**

\[
RMSE = \sqrt{MSE}
\]

RMSE is expressed in the same unit as the target variable.

**R² — Coefficient of Determination**

\[
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
\]

General interpretation:
- `1`: Perfect prediction.
- `0`: Equivalent to predicting the target mean, under the standard R² definition.
- Negative: Can be worse than the mean-prediction baseline.

### 9.2 Classification Metrics

A confusion matrix uses four terms:

- **TP (True Positive):** Actual positive, predicted positive.
- **TN (True Negative):** Actual negative, predicted negative.
- **FP (False Positive):** Actual negative, predicted positive.
- **FN (False Negative):** Actual positive, predicted negative.

**Accuracy**

\[
Accuracy = \frac{TP+TN}{TP+TN+FP+FN}
\]

**Precision**

\[
Precision = \frac{TP}{TP+FP}
\]

**Recall**

\[
Recall = \frac{TP}{TP+FN}
\]

**F1-score**

\[
F1 = 2\times\frac{Precision\times Recall}{Precision+Recall}
\]

- **Accuracy:** The proportion of all predictions that are correct.
- **Precision:** Among predicted positives, the proportion that are actually positive.
- **Recall:** Among actual positives, the proportion correctly identified.
- **F1-score:** The harmonic mean of Precision and Recall.

**Note:** For imbalanced datasets, accuracy alone may be misleading. Consider class distribution, the confusion matrix, Precision, Recall, and F1-score.


## 10. Data Preprocessing and Feature Scaling

Before training, data may need to be cleaned or transformed. This process is called **Data Preprocessing**.

| Concept | Purpose |
|---|---|
| Missing Value Handling | Handle missing data |
| Encoding | Convert categorical data into numerical representations |
| Normalization | Scale values to a specified range |
| Standardization | Scale features using their mean and standard deviation |
| Feature Selection | Select relevant input features |
| Data Cleaning | Handle duplicates and incorrect values |

### Min-Max Scaling

\[
X' = \frac{X-X_{min}}{X_{max}-X_{min}}
\]

This commonly transforms values into the range 0 to 1.

### Standardization

\[
Z = \frac{X-\mu}{\sigma}
\]

Where:
- `μ` = Mean
- `σ` = Standard deviation

After standardization, a feature generally has a mean of 0 and a standard deviation of 1.

Feature scaling is especially important for KNN, SVM, and many distance-based or scale-sensitive algorithms. It is not equally necessary for every algorithm.


## 11. Data Leakage

**Data Leakage** occurs when information that should not be available during model training or prediction influences the training process or evaluation.

Example:
- Fit a scaler using the entire dataset.
- Split the dataset into training and test sets afterward.

The preprocessing step has already used information from the test set. This can make evaluation results overly optimistic.

### Safer Workflow

1. Split the data into training and test sets first.
2. Fit preprocessing steps using only the training data.
3. Transform both training and test data using the fitted preprocessing steps.
4. Train the model using training data only.
5. Use the test set for final evaluation.

Scikit-learn's `Pipeline` can combine preprocessing and modeling steps and help prevent leakage, especially during cross-validation.

# Practical Lab 01: Iris Flower Classification

## Objective

Build, train, and evaluate a Decision Tree Classifier using the built-in Iris dataset from Scikit-learn.

### Dataset Information

The Iris dataset contains:
- 150 flower samples
- 4 numerical features
- 3 species classes: Setosa, Versicolor, and Virginica

Features:
1. Sepal length
2. Sepal width
3. Petal length
4. Petal width

This is a **multiclass classification** problem.


```python
# Step 1: Import the required libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

print("Libraries imported successfully!")
```

```python
# Step 2: Load the Iris dataset

iris = load_iris()

X = iris.data
y = iris.target

df = pd.DataFrame(X, columns=iris.feature_names)
df["species"] = y
df["species_name"] = [
    iris.target_names[label] for label in y
]

display(df.head())
```

```python
# Step 3: Explore the dataset

print("Dataset shape:", X.shape)
print("Number of samples:", X.shape[0])
print("Number of features:", X.shape[1])
print("Target classes:", iris.target_names)

print("\nDataset information:")
df.info()

print("\nStatistical summary:")
display(df.describe())

print("\nClass distribution:")
print(df["species_name"].value_counts())
```

```python
# Step 4: Visualize two features

plt.figure(figsize=(8, 5))

scatter = plt.scatter(
    X[:, 0],
    X[:, 1],
    c=y,
    cmap="viridis",
    edgecolor="black",
    alpha=0.8
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Iris Dataset: Sepal Length vs. Sepal Width")
plt.colorbar(scatter, label="Class")
plt.grid(alpha=0.2)
plt.show()
```

### Visualization Explanation

- `X[:, 0]` selects the first feature (sepal length) for every sample.
- `X[:, 1]` selects the second feature (sepal width) for every sample.
- `c=y` colors each point according to its target class.
- A scatter plot helps us inspect the distribution of samples and classes.

Only two features are shown in this visualization. The model will use all four features during training.


```python
# Step 5: Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Training feature shape:", X_train.shape)
print("Testing feature shape:", X_test.shape)
```

**Parameter Explanation:**

- `test_size=0.20`: Uses 20% of the data for testing.
- `random_state=42`: Makes the split reproducible.
- `stratify=y`: Helps preserve class proportions in the training and test sets.

This introductory lab uses a train-test split without a separate validation set.


```python
# Step 6: Create the Decision Tree model

model = DecisionTreeClassifier(random_state=42)

print(model)
```

```python
# Step 7: Train the model

model.fit(X_train, y_train)

print("Model training completed!")
```

```python
# Step 8: Generate predictions

y_pred = model.predict(X_test)

print("Predicted labels:")
print(y_pred)

print("\nActual labels:")
print(y_test)
```

```python
# Step 9: Evaluate the model

accuracy = accuracy_score(y_test, y_pred)

print(f"Test Accuracy: {accuracy:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_pred,
        target_names=iris.target_names
    )
)
```

```python
# Step 10: Display the confusion matrix

cm = confusion_matrix(y_test, y_pred)

cm_display = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=iris.target_names
)

cm_display.plot(cmap="Blues", values_format="d")
plt.title("Iris Classification - Confusion Matrix")
plt.show()
```

```python
# Step 11: Predict the species of a new flower

# Input order:
# sepal length, sepal width, petal length, petal width

new_flower = np.array([[5.1, 3.5, 1.4, 0.2]])

prediction = model.predict(new_flower)
predicted_species = iris.target_names[prediction[0]]

print("Predicted flower species:", predicted_species)
```
