```python
import kagglehub, os

path = kagglehub.dataset_download("anshulm257/rice-disease-dataset")
print(path)

# peek at folder structure
for root, dirs, files in os.walk(path):
    level = root.replace(path, '').count(os.sep)
    print('  ' * level + os.path.basename(root) + '/')
    for f in files[:3]:
        print('  ' * (level+1) + f)
```
